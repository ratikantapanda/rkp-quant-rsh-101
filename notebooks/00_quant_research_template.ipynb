```python
# Example data loading pattern

def load_sample_frame() -> pd.DataFrame:
    idx = pd.date_range(start="2010-01-01", periods=60, freq="M")
    df = pd.DataFrame({
        "signal_a": np.linspace(1, 5, len(idx)),
        "signal_b": np.linspace(5, 1, len(idx)),
        "price": np.cumsum(np.random.default_rng(42).normal(0, 1, len(idx))) + 100,
    }, index=idx)
    return df

sample_df = load_sample_frame()
print(sample_df.head())
print(f"Rows: {len(sample_df)}")
print(f"Columns: {list(sample_df.columns)}")
```

## 2. Data validation
Check missing values, duplicate dates, and unusual noise before building anything.

```python
# Example validation checks

def validate_frame(df: pd.DataFrame) -> None:
    assert not df.empty, "DataFrame is empty"
    assert not df.index.has_duplicates, "Duplicate index values found"
    missing = df.isna().sum()
    if missing.any():
        print("Missing values:\n", missing[missing > 0])

validate_frame(sample_df)
print("Data validation passed.")
```

## 3. Feature and signal idea
Keep the first working idea simple. Example: create a simple signal from a price-derived feature.

```python
# Example signal: rolling z-score of price return

def build_pressure_signal(df: pd.DataFrame) -> pd.Series:
    returns = df["price"].pct_change().fillna(0)
    rolling_mean = returns.rolling(12).mean()
    rolling_std = returns.rolling(12).std().replace(0, np.nan)
    zscore = (returns - rolling_mean) / rolling_std
    return zscore.rename("signal_score")

signal = build_pressure_signal(sample_df)
print(signal.tail())
```
